```python
import matplotlib.pyplot as plt

top20 = consumo_municipio.head(20)

plt.figure(figsize=(12, 6))

top20.sort_values().plot(kind="barh")

plt.title("20 municípios com maior consumo acumulado de energia")
plt.xlabel("Consumo (MWh)")
plt.ylabel("Município")

plt.tight_layout()
plt.show()
```

## Primeiras observações

O ranking mostra que poucos municípios concentram grande parte do consumo de energia elétrica em Santa Catarina.

Joinville apresenta o maior consumo acumulado da série histórica, seguido por Florianópolis e Blumenau. Também aparecem entre os primeiros municípios reconhecidos por sua atividade industrial, portuária ou tecnológica.

Esses resultados sugerem uma concentração do consumo em polos econômicos do estado, hipótese que será aprofundada nas próximas análises.

## Investigação 1.1 — Qual classe de consumo concentra a maior demanda de energia?

```python
consumo_classe = (
    df_long.groupby("classe")["consumo_mwh"].sum().sort_values(ascending=False)
)

consumo_classe
```

```python
plt.figure(figsize=(8, 5))

consumo_classe.sort_values().plot(kind="barh")

plt.title("Consumo acumulado por classe")
plt.xlabel("Consumo (MWh)")
plt.ylabel("Classe")

plt.tight_layout()
plt.show()
```

## Interpretação

A análise por classe mostra que o consumo industrial representa a maior parcela da demanda de energia elétrica em Santa Catarina.

Em seguida aparecem as classes Residencial e Comercial.

Esse resultado é consistente com o perfil econômico do estado, caracterizado por uma forte presença do setor industrial, especialmente em municípios como Joinville, Blumenau, Jaraguá do Sul, Brusque e Chapecó.

Esses resultados reforçam a hipótese de que a atividade econômica exerce papel relevante na concentração do consumo de energia observada entre os municípios.

## Investigação 1.2 — Como o consumo está distribuído entre os mercados Cativo e Livre?


```python
consumo_tipo = df_long.groupby("tipo")["consumo_mwh"].sum().sort_values(ascending=False)

consumo_tipo
```

## Interpretação

O consumo acumulado mostra que o mercado Cativo representa a maior parcela da demanda de energia elétrica em Santa Catarina ao longo da série histórica analisada.

O mercado Livre, embora menor em volume total, também possui participação relevante, indicando a presença de consumidores de maior porte que negociam energia diretamente no Ambiente de Contratação Livre (ACL).

Nas próximas etapas será investigado se essa participação varia entre os municípios e entre as diferentes classes de consumo.

```python
consumo_municipio_tipo = (
    df_long.groupby(["município", "tipo"])["consumo_mwh"].sum().unstack(fill_value=0)
)

consumo_municipio_tipo.head()
```

```python
consumo_municipio_tipo["Livre"].sort_values(ascending=False).head(20)
```

```python
consumo_municipio_tipo["Cativo"].sort_values(ascending=False).head(20)
```

```python
participacao = consumo_municipio_tipo.copy()

participacao["% Livre"] = (
    participacao["Livre"] / (participacao["Livre"] + participacao["Cativo"])
) * 100

participacao.sort_values("% Livre", ascending=False).head(20)
```

```python
plt.figure(figsize=(8, 6))

participacao["% Livre"].sort_values(ascending=False).head(20).plot(kind="barh")

plt.title("20 municípios com maior participação do Mercado Livre")
plt.xlabel("Participação do Mercado Livre (%)")
plt.ylabel("Município")

plt.gca().invert_yaxis()  # coloca o maior no topo

plt.tight_layout()
plt.show()
```

## Conclusão da Pergunta de Pesquisa 1

A investigação mostrou que a distribuição do consumo de energia elétrica em Santa Catarina apresenta diferentes padrões quando analisada sob perspectivas distintas.

A análise do consumo acumulado evidencia que municípios como Joinville, Florianópolis e Blumenau concentram os maiores volumes de energia consumida ao longo da série histórica, refletindo seu peso econômico e populacional.

Entretanto, ao analisar a participação relativa do Mercado Livre (ACL), observa-se um cenário diferente. Municípios de porte médio, como Vidal Ramos, Correia Pinto, Três Barras e Xanxerê, apresentam as maiores participações percentuais do Ambiente de Contratação Livre, indicando a presença de consumidores de grande porte, geralmente ligados às atividades industriais, agroindustriais ou florestais.

Esses resultados demonstram que o volume absoluto de consumo, por si só, não caracteriza completamente o perfil energético de um município. A participação relativa do Mercado Livre surge como um indicador complementar, capaz de revelar diferenças estruturais entre as economias municipais.

Assim, a primeira pergunta de pesquisa pode ser considerada respondida. Conclui-se que o consumo de energia elétrica em Santa Catarina está concentrado em poucos polos econômicos, mas a estrutura desse consumo varia significativamente entre os municípios quando se observa a participação do Mercado Livre.
